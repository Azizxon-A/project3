```python
from sklearn import linear_model

regr = linear_model.LinearRegression()

X = [
    [168, 0], [166, 0], [173, 0], [165, 0], [177, 0], [163, 0],
    [178, 0], [172, 0], [163, 1], [162, 1], [171, 1], [162, 1],
    [164, 1], [162, 1], [158, 1], [173, 1],
]  # input data : 2 dimension

y = [65, 61, 68, 63, 68, 61, 76, 67, 55, 51, 59, 53, 61, 56, 44, 57]  # y:1 dimension

regr.fit(X, y)  # learning

print('coef :', regr.coef_)
print('intercept :', regr.intercept_)
print('regr_score :', regr.score(X, y))
print('Estimated weight of mamur and sevinch :', regr.predict([[167, 0], [167, 1]]))
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn import linear_model

# 1. Данные и обучение
data = {
    'Volume': [1000, 1200, 1100, 1300, 1500, 1400, 1600, 1700, 1100, 1200],
    'Weight': [790, 1160, 929, 865, 1140, 1100, 1280, 1330, 980, 1150],
    'CO2': [99, 95, 90, 99, 105, 107, 112, 115, 92, 98],
}
df = pd.DataFrame(data)

X = df[['Volume', 'Weight']]
y = df['CO2']

model = linear_model.LinearRegression()
model.fit(X, y)

# 2. Увеличенная сетка точек (по 50 значений для каждой оси)
volume_range = np.linspace(df['Volume'].min(), df['Volume'].max(), 50)
weight_range = np.linspace(df['Weight'].min(), df['Weight'].max(), 50)
volume_grid, weight_grid = np.meshgrid(volume_range, weight_range)

# 3. Расчет плоскости
co2_pred = (
    model.intercept_
    + model.coef_[0] * volume_grid
    + model.coef_[1] * weight_grid
)

# 4. Построение 3D-графика
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

# Исходные точки
ax.scatter(
    df['Volume'], df['Weight'], df['CO2'], color='red', s=50, label='Data'
)

# Гладкая плоскость регрессии
ax.plot_surface(
    volume_grid, weight_grid, co2_pred, alpha=0.6, cmap='viridis'
)

ax.set_xlabel('Volume')
ax.set_ylabel('Weight')
ax.set_zlabel('CO2')

# Добавление вашего имени внизу графика
plt.figtext(
    0.5,
    0.02,
    'Abduraxmanov Azizxon',
    ha='center',
    fontsize=12,
    fontweight='bold',
)

plt.show()
```
